```python

frase = input("Introduce una frase: ")
print(frase.upper())
print(frase.lower())
print(frase.capitalize())
```

```python
texto = input("Introduce un texto: ")
texto_modificado = texto.replace('a', 'o')
print(texto_modificado)
```

```python

frase = input("Introduce una frase: ")
lista_palabras = frase.split()
print("Lista de palabras:", lista_palabras)
```

```python
palabra = input("Introduce una palabra: ")
contador =0
for letra in palabra:
    if letra in "aeiouAEIOU":
        contador= contador +1
print(f"La palabra tiene {contador} vocales")
```

```python
cadena = input("Introduce una cadena: ")
print(cadena[::-1])
```

```python
numeros = input("Introduce números separados por comas: ")
lista_numeros = [int(x.strip()) for x in numeros.split(',')]
lista_numeros.sort()
print(lista_numeros)
```
